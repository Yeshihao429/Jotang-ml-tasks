# 实践1-卷积核

## 1.导入必要的库

```python
from PIL import Image # 根据路径打开图像
import torchvision.transforms.functional as F # 转化为tensor的工具，也可以使用torchvision.transforms.ToTensor
import torch
import numpy  as np  # 转化为numpy数组的必备库
```

```python
img = Image.open("./attachment/outer1.png").convert("RGB")
img
```

## 转化成numpy数组

```python
arr = np.array(img)
print(f"---numpy array:---\nshape:{arr.shape}\ndtype:{arr.dtype}\nmax:{arr.max()}\nmin:{arr.min()}")
h, w = 100, 150
print(f"一个像素(位于{h},{w})的RGB值：{arr[h, w]}")
```

## 转化为Tensor

```python
tensor = F.to_tensor(img)
print(f"---tensor:---\nshape:{tensor.shape}\ndtype:{tensor.dtype}\nmax:{tensor.max().item()}\nmin:{tensor.min().item()}")
print(f"一个像素（位于{h}, {w}）的RGB值：{tensor[:, h, w]}") # tensor是C，H，W；numpy数组是H，W，C
```
